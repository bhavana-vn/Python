```python
# merge two data sets
import pandas as pd
data1={
    'name':['a','b','c','d','e'],
    'age':['123','200','300','400','500'],
    'location':['abc','efg','hig','klm','nop'],
    'email':['a@gmail.com','b@gmail.com','c@gmail.com','d@gmail.com','e@gmail.com'],
}
df1=pd.DataFrame(data1)
data2={
    'email':['f@gmail.com','g@gmail.com','h@gmail.com','d@gmail.com','e@gmail.com'],
     'height':[1,2,3,4,5],
    'position':['xyz','xyz','pqr','stu','vwx']
}
df2=pd.DataFrame(data2)
#print(f'{df2} \n{df1}')
pd.concat([df1,df2],axis=1)
```

```python
pd.concat([df1,df2],axis=1)
```

```python
#merge data using left side
pd.merge(df1,df2,on='email',how='left')
```

```python
pd.merge(df1,df2,on='email',how='right')
```
