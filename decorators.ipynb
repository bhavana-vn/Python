```python
IP='hello bhavana good morning'
def convert_upper(fun):
    def wrapper(*a,**b):
        uppercase=fun(*a,**b)
        return uppercase.title()
    return wrapper

def splitter(func):
    def wrapper(*a,**b):
        op=func(*a,**b)
        return op.split()
    return wrapper
    
@splitter       
@convert_upper      
def string(s):
    return s

print(string(IP))
```

```python
def uppercase(func):
    def wrapper(*args,**kwargs):
        uppercase=func(*args,**kwargs)
        return uppercase.title()
    return wrapper
def splitter(func):
    def wrapper(*args,**kwargs):
        split=func(*args,**kwargs)
        return split.split()
    return wrapper

@splitter
@uppercase
def get_string(s):
    return s
string='Welcome to python world'
r=get_string(string)
print(r)  
```

```python
#Open Closed principle
from abc import ABC,abstractmethod
class Paymentprocessor:    
    @abstractmethod
    def pay(self):
        pass        
class CreditCardPayment(Paymentprocessor):
     def __init__(self,amount):
        self.amount=amount
     def pay(self):
         print(f'Processing credit card payment: {self.amount}' )
class DebitCardPayment(Paymentprocessor):
     def __init__(self,amount):
        self.amount=amount
     def pay(self):
         print(f'Processing debit card payment: {self.amount}' )
class PhonepayPayment(Paymentprocessor):
     def __init__(self,amount):
        self.amount=amount
     def pay(self):
         print(f'Processing phonepay payment: {self.amount}' )

credit=CreditCardPayment(10000)
credit.pay()
debit=DebitCardPayment(100)
debit.pay()
phonepay=PhonepayPayment(10000)
phonepay.pay()
							
```

```python
#Lisko substitution Principle
#child class can replace the parent without unexpected behaviour
class Bird:
    def flying(self):
        print('Bird is flying')
class Penguin(Bird):
    def flying(self):
        raise NotImplementedError
        print("Penguin can't fly")
bird=Bird()
bird.flying()
penguin=Penguin()
penguin.flying()
```

```python
#Integration segrigation Principle
from abc import ABC,abstractmethod
class Worker(ABC):
    @abstractmethod
    def work(self):
        pass
    @abstractmethod
    def eat(self):
        pass
class Human(Worker):
    def work(self):
        print('Human is working')
    def eat(self):
        print("Human is eating")
class Robot(Worker):
    def work(self):
        print('Robot is working')
    def eat(self):
        print("Robot Dosn't eat")
 
```

```python
def factorial(n):
    if n<0:
        return n
    elif n==0:
        return 1
    return n* factorial(n-1)
r=factorial(6)
print(r)
            
```

```python
def is_factorial(n):
    fact=1
    for i in range(1,n+1):
        fact*=i
    return fact
op=is_factorial(5)
print(op)
```

```python
def fibonaccie(n):
    if n<0:
        return n
    elif n==0:
        return 0
    elif n==1:
        return 1
    return fibonaccie(n-1)+fibonaccie(n-2)
for i in range(10):
   r=fibonaccie(i)
   print(r)
    
```

```python
a=10

def fibonnaci(n):
    a,b=0,1
    for _ in range(n):
        yield a
        a,b=b,a+b
        
print(list(fibonnaci(a)))
 
```
