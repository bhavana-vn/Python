```python
##finding common elements in two list without using built in
lst1=['a','b','c','a','d',1,2,3,4,2,4,5,3]
lst2=[1,2,3,4,5,3,2,4,5]
common_elemet=[]
for e1 in lst1:
    for e2 in lst2:
        if e1==e2:
            if e1 not in common_elemet:
                    common_elemet.append(e1)
common_elemet
```

```python
#finding common elements in two list using built in
lst1=['a','b','c','a','d',1,2,3,4,2,4,5,3]
lst2=[1,2,3,4,5,3,2,4,5]
common_items=[]
for i in range(len(lst1)):
    for j in range(len(lst2)):
        if lst1[i]==lst2[j]:
            if lst2[j] not in common_items:
                common_items.append(lst2[j])
common_items
```

```python
from collections import Counter
l=[[1,2,3,4,5,6,7,8,9],[1,2,3,4,5,6,7,8,9],1,2,3,4,5,6,7,8,9,[[[1,2,3,4,5,6,7,8,9]]]]
def flatten(lst):
    return [element for item in lst for element in(flatten(item) if isinstance(item,list) else[item])]
r=flatten(l)
r
new_dict=dict(Counter(r))
new_dict
```

```python
lst=[[1,2,3,4,5,6,7,8,9],[1,2,3,4,5,6,7,8,9],1,2,3,4,5,6,7,8,9,[[[1,2,3,4,5,6,7,8,9]]]]
def Item_freq(l,freq=None):
    if freq is None:
        freq={}
    for item in l:
        if isinstance(item,list):
           r=Item_freq(item,freq)
           print(r)
        else:
            freq[item]=freq.get(item,0)+1
    return freq
    
op=Item_freq(lst)
op
```
