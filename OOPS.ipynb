Inherritence:
============
___________________________

Inheritance is nothing but one class aquaires properties(variables) and methods from onother class.

Existing class called Parent/Base/Super class.

New class inheriting parent features called Chaid/Derived/Sub class called .


Uses:
1. code reusability
2. avoid duplicate code
3. Easier maintenance


Single Inheritance:
------------------

one child class inherits from one parent class 

```python
'''Parent-->child (Employee ---> Developer)
#Developer and inherits common Employee '''
class Employee:
    def work(self):
        print("Employee Working")

class Developer(Employee):
    def code(self):
        print("writing code")

d=Developer()
d.work()
d.code()
```

Multiple Inheritence:
--------------------
child:

 parent2
........
. N parents

one child class inherits from multiple parent classes



```python
class Father:
    def money(self):
        print("Fathers money")
class Mother:
    def gold(self):
        print("mothers gold")
class Child(Father,Mother):
    pass
c=Child()
c.money()
c.gold()
```

```python
#Smartphone -->camera features and music features
class Camera:
    def click(self):
        print("capturing the photos")

class Music:
    def play(self):
        print("playing the music")

class Smartphone(Camera, Music):
     pass

s=Smartphone()
s.click()
s.play()
```

Hierarchical Inheritence:
-------------------------

Multiple child classes inherits single parent class properties(variables) and methods.

child1,child2,child3...............childn------->single parent

```python
class Employee:
    def company_policy(self):
        print("company policy")
class HR(Employee):
    	def hire(self):
    						print("hiring the employees")
                                                    
class Developer(Employee):
    	def coad(self):
    						print("developer coading")

class Tester(Employee):
    	def test(self):
    						print("Testing")


hr=HR()
hr.company_policy()
hr.hire()

dev=Developer()
dev.company_policy()
dev.coad()

tester=Tester()
tester.company_policy()
tester.test()
```
