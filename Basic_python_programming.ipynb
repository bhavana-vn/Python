```python
#Write a python program to solve the quadratic equetion
#the stanared form of quadratic equetion is ax^2+bx+c=0 ,,where a,b,c are real numbers(cefficents) and a=!0,(-b+(b**2-4*a*c)**0.5/(2a)) or (-b-(b**2-4*a*c)**0.5/(2a))
import math
#get  the coefficets  from user:
a=float(input("enter value for coefficents a: "))
b=float(input('enter the value for coefficent b: '))
c=float(input('enter the value for coefficent c: '))
#display input
user_inputs={'a':a,'b':b,'c':c}
print(f'User inputs are :{user_inputs}')
#calculating the discriminant value
discriminant=b**2-4*a*c
#check whether the discriminant is pos,neg or 0
if discriminant>=0:
    root1=(-b+math.sqrt(discriminant))/(2*a)
    root2=(-b-math.sqrt(discriminant))/(2*a)
    print(f'Root1: {root1}')
    print(f'Root2: {root2}')
elif discriminant==0:
    #one of the real root is repeated
    root=-b/(2*a)
    print(f'root :{root}')
else:
    #complex root
    real_part=-b/(2*a)
    imagiary_part=math.sqrt(abs(discriminant))/(2*a)
    print(f'root1: {real_part}+{imagiary_part}i')
    print(f'root2: {real_part}-{imagiary_part}i')

  
```

```python
#Write a pytho program to cheeck whether the given number is prime number.
#prime numbers can be only devid by 1 not with other numbers.eg:2,3,5,7,11.....
num=int(input('enter number: '))
flag=False
if num==1:
      print(f'{num} not a prime number: ')
elif num>1:
    for i in range(2,num):
        if(num%i==0):
                flag=True
                break
else:
     flag=False
if(flag):
     print(f'{num} is not a prime number')
else:
      print(f'{num} is prime number')

	
```

```python
#Write a pytho program to dispay all prime numbers upto some range.
start=int(input('enter your starting number: '))
end=int(input('enter your ending number: '))
print(f'prime numbers between {start} and {end}')

for num in range(start,end+1):
    if num>1:
        for i in range(2,num):
            if(num%i)==0:
                break
            else:
                print(num)
```
